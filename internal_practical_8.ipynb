```python

import numpy as np
import random

# ---------------------------------------
# STUDENT STUDY STRATEGY USING Q-LEARNING
# ---------------------------------------

# Available study strategies (Actions)
strategies = [
    "Reading Textbooks",
    "Practicing Questions",
    "Watching Video Lectures",
    "Revision and Mock Tests"
]

# Student knowledge levels (States)
states = [
    "Beginner",
    "Intermediate",
    "Advanced"
]

# Expected marks for each strategy at each
# knowledge level (simulated environment)
marks = np.array([
    [55, 70, 60, 65],   # Beginner
    [65, 85, 70, 80],   # Intermediate
    [70, 90, 75, 95]    # Advanced
])

# Q-table initialization
q_table = np.zeros((len(states), len(strategies)))

# Q-learning parameters
learning_rate = 0.1
discount_factor = 0.9
epsilon = 0.2
episodes = 1000

# Reward function
def get_reward(state, action):
    return marks[state][action]

# Next state transition
def get_next_state(state, action):
    if action == 1 or action == 3:
        return min(state + 1, 2)
    else:
        return state

# ---------------------------------------
# Q-LEARNING TRAINING
# ---------------------------------------

for episode in range(episodes):

    state = 0  # Start as Beginner

    for step in range(5):

        # Epsilon-greedy action selection
        if random.uniform(0, 1) < epsilon:
            action = random.randint(0, 3)
        else:
            action = np.argmax(q_table[state])

        # Get reward
        reward = get_reward(state, action)

        # Find next state
        next_state = get_next_state(state, action)

        # Q-learning update
        old_value = q_table[state][action]

        next_max = np.max(q_table[next_state])

        new_value = old_value + learning_rate * (
            reward + discount_factor * next_max - old_value
        )

        q_table[state][action] = new_value

        state = next_state

# ---------------------------------------
# DISPLAY RESULTS
# ---------------------------------------

print("\n========== Q-LEARNING RESULTS ==========")

print("\nQ-Table:")
print(np.round(q_table, 2))

print("\nBest Study Strategy for Each Knowledge Level:")

for i in range(len(states)):
    best_action = np.argmax(q_table[i])

    print(
        states[i],
        "->",
        strategies[best_action]
    )

# ---------------------------------------
# RECOMMENDATION FOR STUDENT
# ---------------------------------------

print("\n========== STUDENT RECOMMENDATION ==========")

current_state = 0  # Beginner

print("Current Knowledge Level:", states[current_state])

best_action = np.argmax(q_table[current_state])

print("Recommended Study Strategy:",
      strategies[best_action])

print("Expected Marks:",
      marks[current_state][best_action])

print("\nModel Training Completed Successfully!")
```
